```python
# Vanilla Python: For Loop
#
#Here is a practical example of a `for` loop in vanilla Python, iterating through a list of items and printing them out:
# python
# Define a list of programming topics
topics = ["Variables", "Loops", "Functions", "Dictionaries"]

print("Starting my learning loop:")
for index, topic in enumerate(topics, start=1):
    print(f"{index}. Learning about {topic}")
```
